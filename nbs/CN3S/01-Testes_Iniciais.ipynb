## Constantes

```python
#### DADOS DA BACIA ####
BACIA = "Porto Uruaçu"
ÁREA = 34334 # (km2)
R0 = 350

#### PARAMETROS DO MODELO ####
CNI = 7.35
ALFA = 0.2
BETA = 0.0021
k0 = 1
k1 = 0.316
k2 = 0.305
```

## CN3S class

```python
params = CN3SParams()
model = CN3S(params)
```

```python
past_prec = np.array([385.8, 283.2, 84.9]) # Example precipitation values for the last 3 time steps
prec = 287.6

vj = model.vj(past_prec)
cnv = model.cnv(vj)
s = model.s(cnv)
q_up = model.q_up(prec, s)
r1 = model.r1(prec, q_up, r0 = None)
q_low = model.q_low(r1)
r = model.r(r1, q_low)
q_calc_mm = model.q_calc_mm(q_up, q_low)
q_calc_m3s = model.q_calc_m3s(q_calc_mm)

print(f"vj: {vj}")
print(f"cnv: {cnv}")
print(f"s: {s}")
print(f"q_up: {q_up}")
print(f"q_low: {q_low}")
print(f"r: {r}")
print(f"q_calc_mm: {q_calc_mm}")
print(f"q_calc_m3s: {q_calc_m3s}")
```
